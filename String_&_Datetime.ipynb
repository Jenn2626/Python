```python
import pandas as pd
df_sales_data = pd.read_csv('/content/sales_dataset.csv')
```

```python
df_sales_data.info()
```

```python
df_sales_data
```

```python
 # Spit datatype timestamp
 df_sales_data['JoinDate'] = df_sales_data['JoinDate'].str.split().str[0]
 df_sales_data
```

```python
# Convert to datetime
data_col = ['JoinDate','Birthdate','Order_Date']
for col in data_col:
  df_sales_data[col] = pd.to_datetime(df_sales_data[col])
```

```python
df_sales_data
```

```python
diff = df_sales_data['Order_Date'] - df_sales_data['JoinDate']
df_sales_data['diff_day'] = diff.dt.days
df_sales_data
```

```python
df_groupped = df_sales_data.groupby('Order_Date').agg(count_customer =('CustomerID','count')).reset_index()
df_groupped
```

```python
# Calculate previous date
df_groupped['pre_date'] = df_groupped['Order_Date'].shift(1)
# Calculate previous date of customers
df_groupped['pre_date_customer'] = df_groupped['count_customer'].shift(1)
df_groupped['count_diff_customer'] = df_groupped['count_customer'] - df_groupped['pre_date_customer']
df_groupped
```
