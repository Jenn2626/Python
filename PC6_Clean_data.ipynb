```python
# Read data
import pandas as pd
df_data = pd.read_excel('/content/messy_data.xlsx')
```

```python
df_data.info()
df_data.head(5)
```

```python
# Remove non-alphabetic characters from First_Name and Last_Name
# Find which row include that condition
pattern = r'[^a-zA-Z\s]'
df_data[df_data['First_Name'].str.contains(pattern, na=False, regex=True)]
df_data[df_data['Last_Name'].str.contains(pattern, na=False, regex=True)]

# Remove non-alphabetic: any character matching the pattern will be replaced with an empty string
df_data['First_Name'] = df_data['First_Name'].str.replace(pattern,'', regex = True)
df_data['Last_Name'] = df_data['Last_Name'].str.replace(pattern,'', regex = True)

```

```python
# Recheck
pattern = r'[^a-zA-Z\s]'
df_data[df_data['First_Name'].str.contains(pattern, na=False, regex=True)]
df_data[df_data['Last_Name'].str.contains(pattern, na=False, regex=True)]
```

```python
# Format phone numbers to the format nnn-nnn-nnnn Hint:
# For rows containing any non-numeric characters (such as letters), replace them with an empty string.

# Find phone number is non numeric
pattern = r'[^\d]'
df_data[df_data['Phone_Number'].str.contains(pattern, na=False, regex=True)]
```

```python
# Replace non-numeric with empty string
df_data['Phone_Number'] = df_data['Phone_Number'].str.replace(pattern,'', regex = True)
df_data['Phone_Number']
```

```python
# Format them in the style nnn-nnn-nnnn (e.g., 123-178-3720)
pattern_1 = r'(\d{3})(\d{3})(\d{4})'
pattern_2 = r'\1-\2-\3'
df_data['Phone_Number'] = df_data['Phone_Number'].str.replace(pattern_1, pattern_2, regex=True)
df_data['Phone_Number']
```

```python
"""" Create new columns Street_Address, State, and Zip_Code from the existing Street_Address column
Example: 980 Paper Avenue, Pennsylvania, 18503 should be split as:
Street_Address = ‘980 Paper Avenue’
Zip_Code = 18503
State = ‘Pennsylvania’.
If any part is missing, assign None. Hint: Use .str.extract() with regex or .str.split() to separate the address into components."""
df_data.head(5)
```

```python
# Split Address
df_data['New_Address'] = df_data['Address'].str.split(',')
df_data.head(5)
```

```python

df_data['Street_Address'] = df_data['New_Address'].str[0]
df_data['State'] = df_data['New_Address'].str[1]
df_data['Zip_Code'] = df_data['New_Address'].str[2]
df_data.head(5)
```

```python
#   Missing data replace by None
df_data['Street_Address'] = df_data['Street_Address'].fillna('None')
df_data['State'] = df_data['State'].fillna('None')
df_data['Zip_Code'] = df_data['Zip_Code'].fillna('None')
df_data.head(5)
```

```python
"""Standardize the Do_Not_Contact column to only have two values: Y or N
Hint: You can use .map() or .replace() to convert any variations (e.g., yes, no, y, n) to a consistent format."""
df_data.head(5)
df_data.tail(5)
```

```python
# Lower letter
df_data['Do_Not_Contact'] = df_data['Do_Not_Contact'].str.lower()
df_data.head(5)
```

```python
# Standardize the Do_Not_Contact column to only have two values: Y or N  BY replace
df_data['Do_Not_Contact'] = df_data['Do_Not_Contact'].replace(['yes','no','y','n'],['Y','N','Y','N'])
df_data.head(5)
```
