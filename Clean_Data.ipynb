```python
import pandas as pd
df_sales = pd.read_csv('/content/sales_for _clean.csv')
```

```python
df_sales.info()
df_sales.head(5)
```

```python
# Convert to datetime
# Joindate datatype is timestamp, need to split lay date k lay time
df_sales['JoinDate'] = df_sales['JoinDate'].str.split().str[0]
df_sales
```

```python
# Convert to datetime
date_col =['JoinDate','Birthdate','Order_Date']
for col in date_col:
  df_sales[col] = pd.to_datetime(df_sales[col])
df_sales.info()
df_sales.head(5)
```

```python
# Conver to numeric
df_sales['Revenue'] = df_sales['Revenue'].str.replace('$','')
df_sales['Revenue'] = pd.to_numeric(df_sales['Revenue'])
df_sales.head(5)
```

```python
# Convert to string
str_col =['CustomerID','Mobile']
for col in str_col:
  df_sales[col] = df_sales[col].astype(str)
df_sales.head(5)
```

```python
# Convert Mobile
# Split
df_sales['Mobile'] = df_sales['Mobile'].str.split('.').str[0]
df_sales.head(5)
```

```python

# Condition add '84' in Mobile
def clean_mobile(x):
  if x.startswith('84'):
    return x
  else:
    return '84' + x
df_sales['Mobile'] = df_sales['Mobile'].apply(clean_mobile)
df_sales.head(5)
```

```python
#Convert Name
df_sales['Name'] = df_sales['Name'].str.title()
df_sales.head(5)
```

```python
# Convert Product
# Count values product to know how product category
df_sales['Product'].value_counts()
```

```python
# Strip blank
df_sales['Product'] = df_sales['Product'].str.strip()
df_sales['Product'].value_counts()
```

```python
df_sales['Product']= df_sales['Product'].str.title()
df_sales['Product'].value_counts()
```

```python
# Condition logic
def clean_Product(x):
  if x in ['Smart Phone','Smar Tphone']:
    return 'Smartphone'
  elif x in ['Lap']:
    return 'Laptop'
  elif x in ['Table']:
    return 'Tablet'
  else:
    return x
df_sales['Product'] = df_sales['Product'].apply(clean_Product)
df_sales['Product'].value_counts()
```

```python
df_sales.head(5)
```

```python
# Create find missing data function
def find_missing_value(df):
  missing_data = df.isnull().sum().to_frame('count_null')
  missing_data['percent_null'] = missing_data['count_null']/len(missing_data)
  return missing_data
find_missing_value(df_sales)
#
```
