```python
import pandas as pd
df_customers =pd.read_csv('/content/customers.csv')
df_orders_items = pd.read_csv('/content/order_items.csv')
df_orders = pd.read_csv('/content/orders.csv')
df_products = pd.read_csv('/content/products.csv')
df_sellers = pd.read_csv('/content/sellers.csv')
df_geolocation = pd.read_csv('/content/geolocation.csv')
df_order_payments = pd.read_csv('/content/order_payments.csv')
df_order_reviews = pd.read_csv('/content/order_reviews.csv')
df_product_category_name_translation = pd.read_csv('/content/product_category_name_translation.csv')
#
```

```python
# Condition Logic
def freight_value_category(x):
  if x < 15:
    return 'Cheap'
  elif 15 <= x < 25:
    return 'Medium'
  else:
    return 'Expensive'

df_orders_items['freight_value_category'] = df_orders_items['freight_value'].apply(freight_value_category)
df_orders_items
```

```python
# Lambda:

df_orders_items['freight_value_category'] = df_orders_items['freight_value'].apply(lambda x :'Cheap'if x < 15 else('Midium'if x< 25 else 'Expensive'))
df_orders_items
```

```python
# 2 or more conditions
def discount_category(row):
  if row['price'] > 200:
    if row['freight_value'] > 50:
      return 0.2
    else:
      return 0.1
  else:
    if 100 <= row['price'] <= 200:
      if row['freight_value'] > 50:
        return 0.15
      else:
        return 0.05
  return 0

df_orders_items['category']= df_orders_items.apply(discount_category,axis = 1)
df_orders_items
```

1. Calculate total value = price + freight_value
2. Category total value by mehod:
  total > 500  is hight
  200 < total < 500 is medium
  total < 200 is low
  


```python
df_orders_items['total_value'] = df_orders_items['price'] + df_orders_items['freight_value']

df_orders_items
```

```python
def total_category(x):
  if x > 500:
    return 'High'
  elif 200 < x < 500:
    return 'Medium'
  else:
    return 'Low'

df_orders_items['total_value_category']=df_orders_items['total_value'].apply(total_category)
df_orders_items
```

```python
dfs = [df_product_category_name_translation, df_customers, df_geolocation, df_orders_items, df_orders, df_order_payments,
       df_products, df_order_reviews, df_sellers]
df_names = ['df_product_category_name_translation', 'df_customers', 'df_geolocation', 'df_orders_items', 'df_orders',
            'df_order_payments', 'df_products', 'df_order_reviews', 'df_sellers']

summary_data = []

# Dùng phương thức apply để lấy thông tin cho từng bảng
for df, name in zip(dfs, df_names):
    shape = df.shape  # (số dòng, số cột)
    columns = list(df.columns)  # Tên các cột
    summary_data.append({
        'Tên bảng': name,
        'Shape': shape,
        'Tên cột': columns,
    })

# Tạo DataFrame từ thông tin tổng hợp
summary_df = pd.DataFrame(summary_data)
summary_df
```

# JOIN
1. join giữa các bảng orders, order_items, và customer
2. Tính tổng price và freight_value theo từng customer_state và order_status

```python
import pandas as pd
merge_customer = pd.merge(df_customers, df_orders, how='left', on='customer_id')
merge_customer_order_items = pd.merge(merge_customer, df_orders_items, how='left', on='order_id')

merge_customer_order_items.groupby(['customer_state','order_status']).agg(total_price = ('price','sum'), total_freight_value = ('freight_value', 'sum')).reset_index()
```
